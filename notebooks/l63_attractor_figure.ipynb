## Lorenz-63 attractor figure

Run this notebook to render the Lorenz-63 attractor reconstruction figure.

Rollouts are saved to `figures/figures_inputs/l63_attractor_figure/rollouts.npz`. The figure is written to `figures/l63_attractor/l63_attractor_recon.pdf`.


```python
from __future__ import annotations

import sys
from pathlib import Path

import torch

%matplotlib inline

NOTEBOOK_DIR = Path.cwd().resolve()
REPO_ROOT = NOTEBOOK_DIR.parent if NOTEBOOK_DIR.name == "notebooks" else NOTEBOOK_DIR
SRC_ROOT = REPO_ROOT / "src"
if str(SRC_ROOT) not in sys.path:
    sys.path.insert(0, str(SRC_ROOT))

from experiments.l63_attractor_figure import (
    DEFAULT_SEED,
    export_rollouts,
    plot_l63_attractor_figure,
    resolve_paths,
)
from experiments.lorenz63_main import validate_lorenz63_main_config
from experiments.lorenz63_pipeline import load_config
```

```python
PLOT_CFG = {
    "seed": DEFAULT_SEED,
    "dpi": 200,
    "device": "cpu",
}

config = load_config(REPO_ROOT / "configs" / "lorenz63_main.json")
validate_lorenz63_main_config(config)
device = torch.device(PLOT_CFG["device"])
paths = resolve_paths(REPO_ROOT, output_root=REPO_ROOT / "figures" / "figures_inputs" / "l63_attractor_figure")
FIGURE_DIR = REPO_ROOT / "figures" / "l63_attractor"
CORE_ROOT = REPO_ROOT / "models" / "lorenz63_main_cores"
SHELL_ROOT = REPO_ROOT / "models" / "lorenz63_main_shells"
```

```python
rollouts = export_rollouts(
    config,
    paths,
    core_root=CORE_ROOT,
    shell_root=SHELL_ROOT,
    seed=int(PLOT_CFG["seed"]),
    device=device,
    skip_existing=paths.rollouts_npz.is_file(),
)
print(f"rollouts: {rollouts.relative_to(REPO_ROOT)}")
```

```python
out_pdf = plot_l63_attractor_figure(paths, figure_dir=FIGURE_DIR, dpi=int(PLOT_CFG["dpi"]))
print(f"wrote {out_pdf.relative_to(REPO_ROOT)}")
```
