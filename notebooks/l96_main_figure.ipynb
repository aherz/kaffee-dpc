## Lorenz-96 Experiment Main Figure

Run this notebook to load `figures/figures_inputs/lorenz96_main_figure/` and render the figure. Inputs are committed under that path; regenerate with
```bash
python scripts/run_l96_main_figure.py --device cuda
```

Figures are written to `figures/l96_main/`.


````python
from __future__ import annotations

import sys
from pathlib import Path

import matplotlib.pyplot as plt

%matplotlib inline

NOTEBOOK_DIR = Path.cwd().resolve()
REPO_ROOT = NOTEBOOK_DIR.parent if NOTEBOOK_DIR.name == "notebooks" else NOTEBOOK_DIR
SRC_ROOT = REPO_ROOT / "src"
if str(SRC_ROOT) not in sys.path:
    sys.path.insert(0, str(SRC_ROOT))

from experiments.l96_main_figure import (
    DEFAULT_FIGSIZE,
    DEFAULT_FIGURE_DPI,
    TRACE_STYLE,
    draw_l96_main_figure,
    load_l96_main_figure_data,
    make_l96_main_figure_layout,
    plot_reassim_rollout_panel,
    resolve_paths,
    save_l96_main_figure,
)
````

````python
PLOT_CFG = {
    # State coordinate shown in panel (g); 0–19 for the d=20 system.
    "rollout_dim": 0,
    # ±sigma uncertainty bands in panel (g).
    "uncertainty_sigma": 2.0,
    # Figure size (width, height) in inches.
    "figsize": DEFAULT_FIGSIZE,
    # PNG export resolution.
    "dpi": DEFAULT_FIGURE_DPI,
    # Output stem (writes <stem>.pdf and <stem>.png).
    "stem": "l96_results",
}

paths = resolve_paths(REPO_ROOT)
FIGURE_DIR = REPO_ROOT / "figures" / "l96_main"
````

````python
data = load_l96_main_figure_data(paths, rollout_dim=PLOT_CFG["rollout_dim"])
print(f"trace rows: {len(data.trace_summary)}")
print(f"rollout coordinate: {data.rollout_dim}, stride: {data.rollout_stride}, dt: {data.dt}")
````

### Export the figure


````python
with plt.rc_context(TRACE_STYLE):
    fig, axes = make_l96_main_figure_layout(figsize=PLOT_CFG["figsize"])
    draw_l96_main_figure(
        fig,
        axes,
        data,
        uncertainty_sigma=PLOT_CFG["uncertainty_sigma"],
    )
    out_pdf = save_l96_main_figure(
        fig,
        FIGURE_DIR,
        stem=PLOT_CFG["stem"],
        dpi=PLOT_CFG["dpi"],
    )
print(f"wrote {out_pdf.relative_to(REPO_ROOT)}")
````
