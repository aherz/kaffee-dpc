## DynaMix Experiment Main Figure

Run this notebook to load `figures/figures_inputs/dynamix_main_figure/` and render the main DynaMix figure from the paper.

Regenerate inputs from `results/dynamix_main/adaptations` via
```bash
python scripts/run_dynamix_main_figure.py --device cpu
```
Figures are written to `figures/dynamix_main/dynamix_context_adaptation`.


````python
from __future__ import annotations

import sys
from pathlib import Path

import matplotlib.pyplot as plt

%matplotlib inline

NOTEBOOK_DIR = Path.cwd().resolve()
REPO_ROOT = NOTEBOOK_DIR.parent if NOTEBOOK_DIR.name == "notebooks" else NOTEBOOK_DIR
SRC_ROOT = REPO_ROOT / "src"
if str(SRC_ROOT) not in sys.path:
    sys.path.insert(0, str(SRC_ROOT))

from experiments.dynamix_main_figure import (
    DEFAULT_FIGSIZE,
    draw_dynamix_main_figure,
    load_dynamix_main_figure_data,
    make_dynamix_main_figure_layout,
    resolve_paths,
    save_dynamix_main_figure,
)
````

````python
PLOT_CFG = {
    "figsize": DEFAULT_FIGSIZE,
    "dpi": 360,
}

paths = resolve_paths(REPO_ROOT)
FIGURE_DIR = REPO_ROOT / "figures" / "dynamix_main"
````

````python
data = load_dynamix_main_figure_data(paths)
print(f"panel (a) rows: {len(data.dpc_panel)}")
print(data.dpc_panel.groupby('method').size())
````

````python
fig, axes = make_dynamix_main_figure_layout(figsize=PLOT_CFG["figsize"])
draw_dynamix_main_figure(fig, axes, data)
out_pdf = save_dynamix_main_figure(fig, FIGURE_DIR, dpi=PLOT_CFG["dpi"])
print(f"wrote {out_pdf.relative_to(REPO_ROOT)}")
````
